```python
best_models_list = ['yolov8s', 'yolo11m']

for model_name in best_models_list:
    model = YOLO(model_name)
    run_name = f"best_{model_name}_runs"

    print('\n'+ '='*40)
    print(f'Start Training model : ({model_name}) ...')
    print('='*40, '\n')
    
    model.train(
        data=data_path,
        epochs=200,   
        imgsz=640,
        patience=50,
        batch=16,
        exist_ok=True,
        optimizer="AdamW",
        lr0=0.001,
        weight_decay=0.0002,
        warmup_epochs=5,
        cos_lr=True,
        box=5,
        cls=3,
        name=run_name
    )

print("\nAll models have finished training! Processing comparison charts...")
print("="*50)

!yolo detect predict model=/kaggle/working/runs/detect/train/weights/best.pt \
                     source={ROOT_DATASET}/images/test \
                     save=true
```

```python
rows = []
for model_name in ["yolov8s", "yolo11m"]:
    run_name = f"best_{model_name}_runs"
    csv_path = f"runs/detect/{run_name}/results.csv"
    
    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        df.columns = df.columns.str.strip() # clean white spaces from columns
        
        rows.append({
            "Model": model_name,
            "Train Box Loss": df["train/box_loss"].iloc[-1],
            "Train Class Loss": df["train/cls_loss"].iloc[-1],
            "mAP50 (B)": df["metrics/mAP50(B)"].iloc[-1],
            "mAP50-95 (B)": df["metrics/mAP50-95(B)"].iloc[-1]
        })

comparison_df = pd.DataFrame(rows)
print("\n=== Final Performance Metrics Summary ===")
display(comparison_df)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(20, 14))

for idx, model_name in enumerate(["yolov8s", "yolo11m"]):
    run_name = f"best_{model_name}_runs"
    img_path = f"runs/detect/{run_name}/results.png"
    
    if os.path.exists(img_path):
        img = cv2.imread(img_path)
        axes[idx].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        axes[idx].set_title(f"Charts: {model_name}", fontsize=14)
        axes[idx].axis('off')

plt.tight_layout()
plt.show()
```
